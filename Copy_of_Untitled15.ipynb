```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
ds=pd.read_csv('sample_data.csv')
ds.head()
```

```python
ds.info()
```

```python
ds.isnull().sum()
```

```python
ds.columns
```

```python
y=ds.pop('TARGET CLASS')
x=ds
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
decision_tree=DecisionTreeClassifier()
```

```python
decision_tree.fit(x_train,y_train)
```

```python
predictions=decision_tree.predict(x_test)
```

```python
df=pd.DataFrame({'Actual Values':y_test,'Predicted Values':predictions})
df
```

```python
print(accuracy_score(y_test,predictions))
```

```python
sns.scatterplot(x='Actual Values',y='Predicted Values',data=df)
plt.title("Decision tree classifier's performance")
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.show()
```
